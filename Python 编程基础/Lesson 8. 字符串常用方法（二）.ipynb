## 1. partition 方法

&emsp;&emsp;partition 方法和之前讲的 split 方法有以下区别：

- 其一是 partition 方法不会删除指定切分的元素；
- 其二是 partition 方法在切分包含多个指定切分字符的字符串时，只会针对**第一个字符**进行切分（从左切分）；
- 其三是 partition 方法返回元组，而 split 方法返回列表。

**Point:** partition 方法永远返回具有 3 个元素的元组！

```python
# 以 e 为切分字符，切分字符串 Hello。
"Hello".partition('e')
```

```python
# 只以第一个 e 为切分字符
s = "Hello"
(s*2).partition('e')
```

提示：partition 方法返回结果的对象类型为元组，该类型对象会在后续课程中进行讲解。

```python
# 切分字符出现多次时，只针对第一次出现位置字符进行切分。
"Hello".partition('l')
```

```python
# 切分首、尾字符时仍然会出现空字符
"Hello".partition('H')
```

```python
"Hello".partition('o')
```

**Point:** rpartition 方法（从右切分）

```python
"Hello".rpartition('l')
```

## 2. upper & lower 方法

&emsp;&emsp;顾名思义，upper 方法就是将字符串中的字符全部改为大写，lower 方法则是将字符全都转化为小写。

```python
"Hello".upper()
```

```python
"Hello".lower()
```

## 3. strip 方法

&emsp;&emsp;用于去除字符串首尾指定字符，当参数为空时，默认删除首尾空格。

```python
"Hello  ".strip()
```

```python
" Hello ".strip()
```

```python
# 也可删除指定字符
"HHello".strip('He')
```

```python
"HHelloH".strip('oH')
```

```python
# 只能删除首尾字符，e 字符在字符串内部，无法删除。
"Hello".strip('e')
```

## 4. join 方法

&emsp;&emsp;用于在指定字符串中间穿插其他字符串。

```python
" ".join("Hello")
```

## 5. format 方法

&emsp;&emsp;Python 字符串的 format 方法主要用于格式化输出，应用较广，但大多数场景和数据分析无关，此处只列举部分应用场景进行讲解。

- 大括号对应

```python
url = "http://www.{}.com"
```

```python
# 可实现针对大括号内进行字符串的自动填充
url.format('baidu')
```

```python
url.format('sina')
```

- 数字对应

```python
# 多位置可利用数字进行自动填充
text = "{2}，你好！欢迎使用 {0} 系统，请登录 {1} 获取更详细的信息！"
```

```python
text.format('AIOps', 'https://github.com', 'Alberthua')
```

- 关键字对应

```python
text = "{姓名}，你好！欢迎使用 {平台} 系统，请登录 {地址} 获取更详细的信息！"
```

```python
text.format(姓名="Alberthua", 平台='AIOps', 地址='https://github.com')
```

## 6. center 方法

&emsp;&emsp;通过在字符串两边添加指定的字符使字符串居中显示，默认为空格。

```python
# 返回的字符串总共占 50 个字符，居中显示并用空格两边填充。
"Welcome to use AIOps system!".center(50)
```

```python
# 返回的字符串总共占 40 个字符，居中显示并用 * 号两边填充。
"Welcome to use AIOps system!".center(40, '*')
```
