## 1. 集合元素增加、删除方法

### 1.1 add 方法

&emsp;&emsp;一种非常常见的用于在集合中添加元素的方法。

```python
s1 = {1, 3, 2}
```

```python
print(s1)
```

```python
# 在 s1 集合中添加元素 4
s1.add(4)
```

```python
print(s1)
```

### 1.2 remove 方法

&emsp;&emsp;常用于删除集合中的指定元素，由于集合本身是无序对象，且元素没有 key-value 二元属性，因此只能通过指定的方法对元素进行删除。

```python
print(s1)
```

```python
s1.remove(4)
```

```python
s1
```

```python
s2 = {(1, 2), 'Hi', 1, 2}
```

## 2. 集合的数学运算

### 2.1 交集运算

```python
# 准备集合
s1 = {1, 2, 3}
s2 = {2, 3, 4}
```

```python
# 使用运算符 & 进行交集运算
s1 & s2
```

```python
# 使用 intersection 方法进行交集运算
s1.intersection(s2)
```

```python
s2.intersection(s1)
```

```python
# 使用 intersection_update 方法进行交集运算并覆盖原值
s1.intersection_update(s2)
```

```python
s1
```

### 2.2 并集运算

```python
# 准备集合
s1 = {1, 2, 3}
s2 = {2, 3, 4}
```

```python
# 利用运算符 | 进行并集运算
s1 | s2
```

```python
# 利用 union 方法进行并集运算
s1.union(s2)
```

```python
s1
```

```python
# 使用 update 进行并集运算并更新原集合
s1.update(s2)
```

```python
s1
```

### 2.3 差集运算

```python
# 准备集合
s1 = {1, 2, 3}
s2 = {2, 3, 4}
```

```python
# 使用 - 运算符进行差集计算
s1 - s2
```

```python
s2 - s1
```

```python
# 使用 difference 方法求差集
s1.difference(s2)
```

```python
# 使用 difference_update 方法求差集并更新原集合
s1.difference_update(s2)
```

```python
s1
```

### 2.4 补集运算

```python
# 准备集合
s1 = {1, 2, 3}
s2 = {2, 3, 4}
```

```python
# 并集
s1 | s2
```

```python
# 交集
s1 & s2
```

```python
(s1 | s2) - (s1 & s2)  # 并集与交集的差集为补集
```

```python
# 利用运算符 ^ 进行补集运算
s1 ^ s2
```

```python
s2 ^ s1
```

```python
# 利用 symmetric_difference 方法进行补集运算
s1.symmetric_difference(s2)
```

```python
s2.symmetric_difference(s1)
```

```python
# 利用 symmetric_difference_update 方法进行补集运算并更新原集合
s1.symmetric_difference_update(s2)
```

```python
s1
```

### 2.5 集合关系判断

```python
# 准备集合
s1 = {1, 2, 3}
s2 = {2, 3}
```

```python
# 使用 issubset 方法判断是否为子集
s2.issubset(s1)
```

```python
# 使用 issuperset 判断是否为超集合
s1.issuperset(s2)
```

```python
# 使用 isdisjoint 判断是否包含相同元素，若有相同元素，则返回 False。
s1.isdisjoint(s2)
```

```python
s3 = {1, 2}
s4 = {3}
```

```python
s3.isdisjoint(s4)
```
