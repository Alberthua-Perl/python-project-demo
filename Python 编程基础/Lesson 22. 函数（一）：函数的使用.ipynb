## 1. 引子

&emsp;&emsp;之前我们谈到，判别表达式和循环表达式，都是除了代码顺序执行外的一些常用组织 Python 世界中各个元素的基本逻辑，这些逻辑最终都是为了完成某些功能的实现，例如，想要比较一系列数值（1, 2, 5, 7）的大小，并挑选出其中最大的一个，则可以通过以下方式实现：

Step 1. 将这些数放入列表容器中，或将这些数储存在一个列表中：

```python
l = [-2900, 1, 2, 5, 82.903, 7, -3.86, 6]
```

Step 2. 通过判别循环，找出最大的元素：

```python
ma = l[0]

for i in range(1, len(l)):
    if ma < l[i]:
        ma = l[i]
    else:
        continue

print(f"最大值：{ma}")
```

&emsp;&emsp;通过一个赋值语句，以及一个循环语句，就能够顺利的找出一组数中的最大值。此时，若想复用该逻辑，即再针对另一列数找出最大值，如（22, 51, 19, 43），我们可以再写一遍上述代码：

```python
l1 = [22, 51, 19, 43]
ma1 = l1[0]

for i in range(1, len(l1)):
    if ma1 < l1[i]:
        ma1 = l1[i]
    else:
        continue

print(f"最大值：{ma1}")
```

&emsp;&emsp;好像也不是很复杂，但若逻辑进一步复杂化，例如想进一步找出这组数中的最小值，应该怎么办呢？

Step 3. 通过循环判别，找出最小元素：

```python
mi = l[0]

for i in range(1, len(l)):
    if mi > l[i]:
        mi = l[i]
    else:
        continue

print(f"最小值：{mi}")
```

&emsp;&emsp;再进一步，假设我现在想让该列数中的每个元素都减去该列数的最小值，再除以该列数的极差（最大值-最小值），该如何操作？

Step 4. 计算极差

```python
mr = ma - mi
```

```python
mr
```

Step 5. 利用循环，按要求处理每个数据

```python
for i in range(len(l)):
    #l[i] = round((l[i] - mi) / mr, 6)
    l[i] = (l[i] - mi) / mr
    print(f"第 {i} 个元素标准化：{l[i]:.6f}")
```

```python
l
```

&emsp;&emsp;按照此逻辑进行处理，本质上就是对这组数进行了 0-1 标准化处理，这是数据分析中最常见的一种对数据处理的 **归一化方法**，也是比较简单的处理数据的方法之一。但尽管如此，上述逻辑如果要复用，可能仍然会比较复杂，有没有什么简单的方法可以使得复用逻辑变得简单一些呢？有！就是利用函数。

## 2. 函数功能

&emsp;&emsp;从使用角度来说，函数的本质就是对现有的逻辑进行封装，以方便以后再次调用、将相同的逻辑用于处理其他对象。例如，我们可以将上述归一化的逻辑封装为如下函数：

&emsp;&emsp;首先，我们再来看下之前的逻辑能否进一步合并简化：

```python
l = [-2900, 1, 2, 5, 82.903, 7, -3.86, 6]
```

```python
ma = l[0]
mi = l[0]

for i in range(1, len(l)):
    if ma < l[i]:
        ma = l[i]
    if mi > l[i]:
        mi = l[i]

mr = ma - mi
```

```python
mr
```

```python
l
```

&emsp;&emsp;再封装为一整个函数：

```python
# 完整函数示例
def min_max_norm(l: list):
    """
    功能：归一化输入列表各元素的值，即列表中的每个元素减去列表的最小值除以列表最大值与最小值的差。
    参数：形参 l 为列表对象
    返回：归一化后的列表对象
    """
    ma = l[0]
    mi = l[0]
    for i in range(1, len(l)):
        if ma < l[i]:
            ma = l[i]
        if mi > l[i]:
            mi = l[i]
    mr = ma - mi
    for i in range(len(l)):
        l[i] = (l[i] - mi) / mr
    return l
```

&emsp;&emsp;尝试着调用函数：

```python
min_max_norm(l)
```

&emsp;&emsp;尝试针对其他对象复用该逻辑：

```python
l1 = [22, 51, 19, 43]
l1
```

```python
min_max_norm(l1)
```

## 3. 函数结构详解

&emsp;&emsp;函数的定义其实并不复杂，总的来说分为两部分，其一是通过 **def 关键词** 定义函数名称及参数，第二部分则是具体代码部分。例如，在上述函数中，我们通过 `def min_max_norm(l: list):` 语句实际上定义了该函数的名称 min_max_norm，以及该函数在实际执行过程中涉及到的对象，即 **函数参数**。其中，函数名称取名要求尽量简洁易懂，同时和内嵌函数名称不能重复，而函数参数，其功能类似于方程组中变量 $x$、$y$ 的作用，表示函数体内将要用到的对象名，本质上起到 **占位** 的作用，而函数代码则用于表示函数参数所代表的对象实际操作逻辑，因此，函数本质上其实是处理逻辑的封装。例如：

```python
def sum_1(a, b):    # 函数名为 sum_1，a、b 为函数参数（形参）。
    c = a + b       # 该函数实现的操作逻辑是让 a、b 相加，并且将结果赋给变量 c。
    return c        # 函数最终返回 c
```

```python
sum_1(1, 2)
```

```python
sum_1([1, 2], [3, 4])    # 列表对象连接运算符
```

&emsp;&emsp;函数中的 **return 关键字** 表示函数最终返回的对象，函数执行完之后，我们可以将函数返回对象赋给某变量，当然也可以直接在屏幕上打印然后释放该对象所占用内存。

## 4. 函数的调用过程

&emsp;&emsp;函数一旦定义完毕后，就可以调用使用。**调用函数的过程本质上就是给函数参数（占位符）进行赋值，或者说尝试使用函数内部定义的逻辑去操作给定的对象。** 例如，之前的 sum_1 函数，该函数封装的逻辑是将两个对象进行运算符加号的计算，然后返回计算结果，则当执行：

```python
sum_1(1, 2)
```

时，本质上是进行了

```python
1 + 2
```

的计算。同理，当执行:

```python
sum_1([1, 2], [3, 4])
```

时，实际上是进行了

```python
[1, 2] + [3, 4]
```

的计算。因此，定义函数的本质是封装计算逻辑，而调用函数的本质则是将这些计算逻辑作用在确定的对象上。

## 5. 函数帮助文档

- 作为操作逻辑的封装，为了保证使用的便捷性，常常在定义函数的时候会对函数功能进行说明，这就是所谓的函数帮助文档。
- 自定义函数帮助文档格式如下：

```python
def sum_1(a, b):
    """
    让输入的两个对象利用加号运算符进行相加
    参数：
      a：函数输入的第一个对象
      b：函数输入的第二个对象
    return，返回 a+b 的结果。
    """
    c = a + b
    return c
```

&emsp;&emsp;合理使用帮助文档，能大幅提高函数的易用性。此时，若使用者知道是利用加号进行操作，则可基于对加号运算符的理解来使用函数，例如：

```python
sum_1(True, True)    # True 相当于 1，False 相当于 0。
```
