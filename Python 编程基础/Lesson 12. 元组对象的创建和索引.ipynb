# Lesson 12. 元组对象的创建和索引&emsp;<span style="color:blue">[DONE]</span>

&emsp;&emsp;元组（tuple）是 Python 中比较抽象的对象类型，是依据计算机代码执行的功能特点衍生的一类对象。使用上和列表类似，都有装载其他对象的容器功能，并且无论是对象创建方法，还是对应的对象常用方法都非常类似。但二者又拥有本质上的不同，其中最核心的不同就表现在 **元组属于不可变对象类型**，由此很多基于列表的增删改查相关操作对元组来说都无法执行，并且正是由于其不可变性，在某些情况下，适当使用元组对象将很大程度上确保代码运行的稳定性。

## 1. 元组的创建

和列表不同，Python 中一般使用小括号来标注元组。

### 1.1 一般创建方法

```python
# 使用小括号来创建元组
t = (1, 2, 3)
```

```python
# Python 中 tuple 表示元组
type(t)
```

```python
print(t)
```

Q: 如何创建一个只包含一个元素的元组？

```python
(1)
```

```python
# 此对象不是元组，只是整数
type((1))
```

```python
# 正确的创建只包含一个元素元组的方法
(1, )
```

```python
type((1, ))
```

Q: 如何创建一个空元组呢？

```python
# 空元组和空列表创建方法一样，限定符内不添加任何元素即可。
()
```

```python
type(())
```

元组和列表一样，都拥有**容器特性**，只不过元组是不可变容器。

```python
# 可存放不同类型对象
('Hi', 1, print, (1, 2), [1, 2])
```

### 1.2 函数创建法

和列表类似，元组也可通过 tuple 函数来进行创建。

```python
# 传入列表，创建元组，也可视作列表转化为元组。
tuple([1, 2])
```

```python
# 若要创建只包含一个元素的元组，也需要将其放入列表中。
tuple([1])
```

```python
# 传入字符串，创建元组，也可理解为将字符串转化为元组。
tuple('Hello')
```

```python
# 元组和列表也可以相互转化
list(tuple('Hello'))
```

### 1.3 借助 range 函数创建元组

和列表一样，当需要创建具有一定分布规律数值的元组时，也可借助 range 函数来实现，其实现过程与创建有规律分布列表方法完全一致。

```python
range(5)
```

```python
tuple(range(5))
```

```python
tuple(range(2, 5))
```

```python
tuple(range(2, 5, 2))
```

```python
tuple(range(5, 2, -2))
```

## 2. 元组的索引和切片

元组的索引切片的方法和列表完全一致。

```python
t1 = tuple(range(5))
print(t1)
```

```python
# 单值索引
t1[2]
```

```python
# 反向索引
t1[-1]
```

```python
# 切片
t1[1: 3]
```

```python
# 反向切片
t1[:: -2]
```

## 3. 元组不可变性讨论

基本理解：元组中包含的元素对象不可变！

```python
t1[1] = 0
```

本质上，元组指向性不变。

```python
'Hello'[1] = 'a'
```

```python
list(t1)
```

```python
# 创建一个包含可变对象的元组
tp2 = ('a', 'b', ['A', 'B'])
print(tp2)
```

```python
# 查看可变对象内容
tp2[2]
```

```python
tp2[2][0]
```

```python
# 修改可变类型对象
tp2[2][0] = 'X'
tp2[2][1] = 'Y'
```

```python
print(tp2)
```

表面上看，tuple 的元素确实变了，但其实改变的不是 tuple 元素，而是 list 元素。tuple 一开始指向的 list 并没有改变成其他别的 list。所以，tuple 所谓的不变是指tuple 中每个元素的指向永远不变。即指向 'a'，就不能改为指向 'b'，指向一个 list，就不能改成指向其他对象，但指向的这个 list 内部元素是可变的。

```python
tp2
```

```python
tp2[2]
```

```python
tp2[2].append('Z')
```
