## 1. 迭代器

- 迭代是 Python 最强大的功能之一，是访问 **集合元素** 的一种方式；
- 迭代器（iterator）是一个可以记住遍历的位置的对象；
- 迭代器对象从集合的第一个元素开始访问，直到所有的元素被访问完结束；
- 迭代器只能往前不会后退；
- 迭代器有两个基本的函数：**iter()** 和 **next()**。

### 1.1 使用迭代器

&emsp;&emsp;字符串，列表或元组对象都可用于创建迭代器：

```python
# 迭代器对象：直接迭代
l = list(range(2, 30, 3))
item = iter(l)           # 创建迭代器对象

for i in item:
    print(i, end=" ")    # print 函数的 end= 参数指定输出值的分隔符
```

```python
# 迭代器对象：next 函数引入
#import sys

l = list(range(2, 30, 3))
item = iter(l)    # 创建迭代器对象

while True:
    try:
        print(next(item))
    except StopIteration:    # 忽略 StopIteration 类错误
        #sys.exit()          # 方法1：退出循环，返回 SystemExit 信息。
        break                # 方法2：直接退出循环
```

### 1.2 创建类的迭代器

- 把一个类作为一个迭代器使用需要在类中实现两个方法 **`__iter__()`** 与 **`__next__()`**。
- 使用 Python 的构造函数 `__init__()`, 它会在对象初始化的时候执行。
- `__iter__()` 方法返回一个特殊的迭代器对象，这个迭代器对象实现了 `__next__()` 方法并通过 **StopIteration** 异常标识迭代的完成。
- `__next__()` 方法（Python 2 里是 next()）会返回下一个迭代器对象。

- 示例：创建一个返回数字的迭代器，初始值为 1，逐步递增 1。

```python
class MyNumbers:
    def __iter__(self):
        self.a = 1
        return self

    def __next__(self):
        x = self.a
        self.a += 3
        return x

myclass = MyNumbers()
myitem = iter(myclass)
print("迭代器对象属性：", myclass.a)

print(next(myitem))
print(next(myitem))
print(next(myitem))
print(next(myitem))
print(next(myitem))
```

### 1.3 StopIteration

- StopIteration 异常用于标识迭代的完成，防止出现无限循环的情况，在 `__next__()` 方法中我们可以设置在完成指定循环次数后触发 StopIteration 异常来结束迭代。
- 示例：在 20 次迭代后停止执行

```python
import sys

class MyNumbers:
    def __init__(self):
        self.a = 1         # 指定对象实例属性
        
    def __iter__(self):    # 迭代器对象
        return self

    def __next__(self):
        x = self.a
        self.a += 1
        if x >= 21:
            print(f"迭代超过 20 次，退出...")
            raise StopIteration    # 结束迭代
            sys.exit()             # 退出运行环境
        else:    
            return x

myclass = MyNumbers()
myitem = iter(myclass)

while True:
    print(next(myitem))
```

## 2. 生成器

- 在 Python 中，使用了 **yield** 的函数被称为生成器（generator）。
- yield 是一个 **关键字**，用于定义生成器函数，生成器函数是一种特殊的函数，可以在迭代过程中逐步产生值，而不是一次性返回所有结果。
- 跟普通函数不同的是，生成器函数是一个返回迭代器对象的函数，只能用于迭代操作，更简单点理解生成器就是一个迭代器。
- 当在生成器函数中使用 yield 语句时，函数的执行将会暂停，并将 yield 后面的表达式作为当前迭代的值返回。
- 然后，每次调用生成器的 next() 函数或使用 for 循环进行迭代时，函数会从上次暂停的地方继续执行，直到再次遇到 yield 语句。这样，生成器函数可以逐步产生值，而不需要一次性计算并返回所有结果。
- **总结：调用一个生成器函数，返回一个迭代器对象。**

- 示例1：生成器函数的作用

```python
def count_down(n):
    while n > 0:
        yield n    # 暂停在此处，返回 n。
        n -= 1

generator = count_down(5)    # 迭代器对象

print(next(generator))
print(next(generator))
print(next(generator))

for x in generator:    # 迭代：迭代器对象
    print(x)
```

&emsp;&emsp;以上示例中，count_down 函数是一个生成器函数。它使用 yield 语句逐步产生从 n 到 1 的倒数数字。在每次调用 yield 语句时，函数会返回当前的倒数值，并在下一次调用时从上次暂停的地方继续执行。<br>
&emsp;&emsp;通过创建生成器对象并使用 next() 函数或 for 循环迭代生成器，我们可以逐步获取生成器函数产生的值。在这个例子中，我们首先使用 next() 函数获取前三个倒数值，然后通过 for 循环获取剩下的两个倒数值。<br>
&emsp;&emsp;**生成器函数的优势是它们可以按需生成值，避免一次性生成大量数据并占用大量内存。** 此外，生成器函数还可以与其他迭代工具（如 for 循环）无缝配合使用，提供简洁和高效的迭代方式。

- 示例2：使用 yield 实现斐波那契数列（即数列中任意一个数是前两个数之和）

```python
import sys, re

def smart_convert(n: str):
    """
    功能：转换字符串为整型数或浮点数
    参数：由数值组成的字符串
    """
    if re.search(r'\.', n):    # 正则匹配字符串中的小数点
        return float(n)
    else:
        return int(n)
        
def fibonacci(seq: list):
    """
    功能：利用生成器函数生成斐波那契数列
    参数：由两个数值对象组成的列表
    """
    if len(seq) < 2:
        print("不满足斐波那契数列初始化要求，退出...")
        sys.exit()
    while len(seq) >= 2:    # 必有循环迭代
        yield seq
        gen = seq[-1] + seq[-2]
        seq.append(gen)

arr = input("请输入初始化数列，形如 a b：")    # 字符串对象
arr_list = [ smart_convert(x) for x in arr.split() ]    # 列表推导式：字符串转换为数值
circle = input("请输入循环次数：")    # 字符串对象
fib = fibonacci(arr_list)

i = 1
for x in fib:
    if i <= int(circle):    # 只要生成循环数内的数组 
        print(f"第{i}个数列：{x}")
    else:
        print("超出循环次数，退出迭代...")    # 跳出循环，防止无限循环，消耗资源。
        break
    i += 1 
```
