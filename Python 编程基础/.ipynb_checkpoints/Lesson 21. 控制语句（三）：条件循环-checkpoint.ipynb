&emsp;&emsp;在上一节的最后，我们看到了一个 if 条件判别和 for 循环结合的例子，在很多实际情况中，都会遇到在一定情况下执行循环的情况，这也就是所谓的条件循环。实现条件循环当然可以利用 for 和 if 的嵌套，但对于大多数情况，我们更推荐使用专门执行条件循环功能的 while 语句。

## 1. while 循环基本结构

&emsp;&emsp;while 循环的基本结构为：

```python
while <判别语句>:
    <表达式>
```

&emsp;&emsp;其中，while 的判别语句与 if 结构中的判别语句一样，当返回结果为 True 时执行表达式，而表达式则是执行循环功能的主体代码，同时，该部分也承担着控制循环是否执行的重要作用。

&emsp;&emsp;例如：创建一个包含三个元素的列表，对前两个元素执行+1的操作。

````python
l = [1, 0, 2]
````

````python
i = 0
````

````python
while i < 2:
    l[i] += 1
    i += 1
````

````python
l
````

````python
i
````

**Points:**
 
- 只要是循环，就必须有实际参与每一次参与循环计算的临时变量，while 循环也不例外，上述例子中 i 就是临时变量，在 while 循环中也被称为 **指示器**；
- 一个比较通用的做法是在循环开始前对临时变量进行赋值，然后临时变量实际参与条件判别，同时在表达式中需要对临时变量进行修改，从而使得条件循环真正意义上有条件可循，否则将会出现 while 条件一直成立，最终造成死循环。

````python
i = 1
sum_i = 0    # 累加操作, 初始值设为0

while i <= 100:
    sum_i += i    # 等价于 sum_i = sum_i + i
    i += 1

print("sum_i 的值是: ", sum_i)
````

&emsp;&emsp;while 循环比较适合 **有序容器**，在有序容器中，可以通过数值型指示器与索引关联来进行有条件的循环。而面对无序变量，则与 for+if 嵌套结构差别不大。同时，while 条件循环一般适用于 “**一直循环直到条件不满足为止**” 的线性条件，如果需要进行更加灵活的条件循环，则需要进一步配合 continue 与 break 语句执行循环。

## 2. 控制结构 continue

&emsp;&emsp;在循环体中，continue 语句功能为结束本次循环进入到下一次循环中。合理利用 continue+if 结构能进行更加灵活的条件循环。例如：

````python
l = [1, 0, 3, 2]
````

&emsp;&emsp;我们想让索引值为奇数元素+1，则：

````python
for i in range(len(l)):
    if i % 2 == 1:
        l[i] += 1
    else:
        continue
````

````python
l
````

&emsp;&emsp;还是对于上述列表，索引值为奇数的+1，为偶数则+2：

````python
l = [1, 0, 3, 2]
l
````

````python
for i in range(len(l)):
    l[i] += 1
    if i % 2 == 1:
        continue
    else:
        l[i] += 1
````

````python
l
````

## 3. 控制结构 break

&emsp;&emsp;和 continue 不同，break 一旦执行，就将终止整个循环，而不是结束本次循环。例如，还是上述列表，需要对前两个元素+1，第三个元素+2，其他元素不变，则：

````python
l = [1, 0, 3, 2]
l
````

````python
for i in range(len(l)):
    if i <= 1:
        l[i] += 1
    elif i == 2:
        l[i] += 2
    else:
        break
````

````python
l
````

````python
l = [1, 0, 3, 2]
l
````

&emsp;&emsp;当然，break 也主要用来表示当触发了某条件后即终止循环，和 while 类似，也只能表示某种线性判别逻辑。

````python
i = 0

while i <= 2:
    l[i] += 1        # l[i] == 4
    if i == 2:
        l[i] += 1    # l[i] == 5
    i += 1
````

````python
l
````
