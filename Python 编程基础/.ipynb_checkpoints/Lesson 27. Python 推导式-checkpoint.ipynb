## 1. 概述

&emsp;&emsp;Python 推导式是一种独特的数据处理方式，可以从一个数据序列构建另一个新的数据序列的结构体。<br>
&emsp;&emsp;Python 推导式是一种强大且简洁的语法，适用于生成列表、元组、字典、集合和生成器。<br>
&emsp;&emsp;在使用推导式时，需要注意可读性，尽量保持表达式简洁，以免影响代码的可读性和可维护性。

&emsp;&emsp;Python 支持各种数据结构的推导式：

- 列表（list）推导式
- 字典（dict）推导式
- 集合（set）推导式
- 元组（tuple）推导式

## 2. 列表推导式

- 列表推导式的基本格式：

  ```plaintext
  [表达式 for 变量 in 列表] 
  [out_exp_res for out_exp in input_list]

  或者 

  [表达式 for 变量 in 列表 if 条件]
  [out_exp_res for out_exp in input_list if condition]
  ```

- out_exp_res：列表生成元素表达式，可以是有返回值的函数。
- for out_exp in input_list：迭代 input_list 将 out_exp 传入到 out_exp_res 表达式中。
- if condition：条件判别语句，可以过滤列表中不符合条件的值。

- 示例1：过滤掉长度小于或等于3的字符串列表，并将剩下的转换成大写字母

````python
names = ["Bob", "Tom", "alice", "Jerry", "Wendy", "Smith"]
all_upper = [ name.upper() for name in names if len(name) >=3 ]
all_upper
````

````python
def trans_first_upper(s):
    """
    功能：转换字符串首字母大写，其余字符全小写。
    参数：
      s：字符串
    """
    inter = list(s.lower())
    inter[0] = inter[0].upper()
    trans_str = ""
    for i in inter: trans_str += i
    return trans_str
````

- 示例2：计算 30 以内可以被 3 整除的整数

````python
numbers = [ num for num in range(31) if num % 3 == 0 ]
numbers
````

## 3. 字典推导式

- 字典推导式的基本格式：

  ```plaintext
  { key_expr: value_expr for value in collection }
  
  或者
  
  { key_expr: value_expr for value in collection if condition }
  ```

- 示例1：使用字符串及其长度创建字典

````python
tup_demo = ("Google", "Tencent", "Taobao")
tup_dict = { key:len(key) for key in tup_demo }
tup_dict
````

- 示例2：提供三个数字，以三个数字为键，三个数字的平方为值来创建字典。

````python
set_num = {5, 19, 37}
set_dict = { num:num ** 2 for num in set_num }
set_dict
````

````python
type(set_dict)
````

## 4. 集合推导式

- 集合推导式的基本格式：

  ```plaintext
  { expression for item in Sequence }

  或者

  { expression for item in Sequence if conditional }
  ```

- 示例1：计算数字 1,2,3 的平方数

````python
set_trans = { num ** 2 for num in (1, 2, 3) }
set_trans
````

- 示例2：判断不是 abc 的字母并输出

````python
alp_list = ["cde", "apple", "abc", "dfe"]
alp_set = { item for item in alp_list if item != "abc" }
alp_set
````

或者

````python
string_set = { alp for alp in "abcdedbcf" if alp not in "abc" }
string_set
````

````python
type(string_set)
````

## 5. 元组推导式（生成器表达式）

- 元组推导式可以利用 range 区间、列表、元组、字典和集合等数据类型，快速生成一个满足指定需求的元组。
- 元组推导式的基本格式：

  ```plaintext
  (expression for item in Sequence )
  
  或者

  (expression for item in Sequence if conditional )
  ```

- 元组推导式和列表推导式的用法也完全相同，只是元组推导式是用 `()` 圆括号将各部分括起来，而列表推导式用的是中括号 `[]`，另外元组推导式返回的结果是一个 **生成器对象**。

- 示例：生成一个包含数字 1~9 的元组

````python
tup_seq = ( item for item in range(1, 10) )
tup_seq    # 返回生成器对象
````

````python
tuple(tup_seq)    # tuple 函数将生成器对象转换为元组
````
