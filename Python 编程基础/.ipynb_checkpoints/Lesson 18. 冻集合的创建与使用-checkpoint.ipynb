# Lesson 18. 冻集合的创建与使用&emsp;<span style="color:blue">[DONE]</span>

&emsp;&emsp;之前我们谈到，集合与字典都是无序的可变类型对象，集合的可变性主要体现在可以通过一些函数或方法直接在原对象基础上进行修改，而与可变集合对应的还有一种不可变类型集合，一般我们将其称为 **冻集合（frozen set）**，其基本性质与集合高度类似，只在可变性上与集合类型不同，**集合可变，而冻集合不可变**。正因如此，冻集合往往可与元组类型对象担当相同的功能，同时，相比集合，由于其不可变性，冻集合只支持常见集合操作，而不支持集合修改操作。

## 1. 冻集合的创建

&emsp;&emsp;冻集合只支持利用函数来进行创建，创建函数为 frozenset。

```python
# 将列表转化为冻集合
frozenset([1, 2, 3])
```

```python
# 将集合转化为冻集合
frozenset({1, 2, 3})
```

```python
# 将元组转化为冻集合
frozenset((1, 2, 3))
```

```python
# 验证冻集合本身不可变性
set([1, 2, frozenset([1, 2])])
```

```python
# 而集合本身是可变的
set([1, 2, {1, 2}])
```

```python
# 冻集合的元素也必须是不可变类型对象
frozenset([1, 2, {1, 2}])  # {1, 2} 为可变对象，因此报错。
```

## 2. 冻集合的常用方法

&emsp;&emsp;由于冻集合本身不可变性，冻集合不支持集合的增、删方法，只支持一些基本集合运算。

```python
fr1 = frozenset([1, 2, 3])
```

冻集合不支持修改：

```python
# 冻集合不允许修改 
fr1.add(1)
```

```python
# 冻集合不允许修改
fr1.pop()
```

```python
fr2 = frozenset([2, 3, 4])
```

&emsp;&emsp;但冻集合支持集合运算：

```python
fr1 & fr2
```

```python
fr1 | fr2
```

```python
fr1 - fr2
```
