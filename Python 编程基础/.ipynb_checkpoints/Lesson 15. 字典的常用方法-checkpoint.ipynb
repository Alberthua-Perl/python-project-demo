## 1. keys/values 方法

&emsp;&emsp;面对字典对象，我们可以使用 keys 或 values 方法选取字典中的 keys 或者 values。

```python
d = dict(A = 1, B = 2, C = 3)
print(d)
```

```python
# values 方法，挑选字典中的 values 组成新的对象。
d.values()
```

```python
# 和 range 类似，返回结果是原对象，若要显式显示其内容，则可使用 list 或者 tuple 函数对其进行转化。
type(d.values())
```

```python
dv = d.values()
```

```python
print(dv)
```

```python
type(dv)
```

```python
# 放入 list 容器
list(dv)
```

```python
# 放入 tuple 容器
tuple(dv)
```

```python
d.keys()
```

```python
list(d.keys())
```

```python
tuple(d.keys())
```

## 2. items 方法

&emsp;&emsp;keys 和 values 方法都是将 key 或者 value 单独返回，而 items 方法则是将每个 key-value 键值对元素打包返回。

```python
# 打包返回键值对，返回结果为由每个键值对构成的元组，并由这些元组所共同构成的列表，该结果仍然属于原类型。
d.items()
```

```python
di = d.items()
```

```python
list(di)
```

```python
tuple(di)
```

**Point:**

实际上，keys/values 也好，items 也好，都是希望能够完成由不可变类型的字典转化为可变类型的转化过程。

## 3. update 方法

&emsp;&emsp;顾名思义，update 方法是用于更新字典的方法，是一种非常灵活的对字典进行修改的方法。

```python
d
```

```python
d1 = dict(B = 2, C = 4, D = 5)
print(d1)
```

字典 d1 相比于字典 d，有一个重复的元素、一个 key 相同但值不同的元素，以及一个新元素，据此测试更新函数的更新效果。

```python
# 将 d1 字典对象中的新的 key-value 键值对更新至 d 字典对象中。
d.update(d1)
```

```python
d
```

**Point:**

- 更新原则是，有新的部分则更换或添加新的部分，其他部分保留；
- 更新是有方向的，d.update(d1) 和 d1.update(d) 有本质上的区别；    
- 更新过程会在原来对象基础上对其进行修改。

```python
d = dict(A = 1, B = 2, C = 3)
d1 = dict(B = 2, C = 4, D = 5)
```

```python
print(d)
```

```python
print(d1)
```

```python
# 使用 d 字典对象更新 d1 字典对象
d1.update(d)
```

```python
print(d1)
```

## 4. setdefault 方法

&emsp;&emsp;setdefault 方法会根据 key-value 键值对原本是否存在，进行有选择的修改。

```python
d = dict(A = 1, B = 2, C = 3)
print(d)
```

setdefault 方法使用过程中，第一个参数位输入备选 key，第二个参数位置输入备选 value。

```python
# 若该备选 key 在原字典中存在，则返回该 key 对应的 value，第二个参数位输入的 value无任何作用。
d.setdefault('A', 5)
```

```python
# 原字典不会被修改
d
```

```python
# 若该备选 key 在原字典中不存在，则返回该 key 对应的 value，备选 key-value 将被添加至字典中。
d.setdefault('D', 5)
```
