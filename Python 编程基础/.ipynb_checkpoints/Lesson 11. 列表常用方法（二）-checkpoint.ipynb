## 1. index 方法

&emsp;&emsp;列表的 index 方法主要用于查找指定内容的索引值。

```python
l = list("Hello")
print(l)
```

```python
# 查找字符 H 在列表中的索引值
l.index('H')
```

```python
# 当查找对象在列表中多次出现时，将返回第一次出现的元素索引值。
l.index('l')
```

```python
# index 方法也可指定索引区间
l.index('l', 3, 5)
```

**Point:** **<span style="color:orange">字符串对象 vs 列表对象</span>**

- 字符串对象使用 find 方法查找指定字符第一次出现的索引值，可搭配切片使用。
- 列表对象使用 index 方法查找指定元素第一次出现的索引值，可搭配切片使用。

## 2. count 方法

&emsp;&emsp;count 方法用于统计指定元素在列表中的出现次数。

```python
l = list("Hello")
print(l)
```

```python
# 统计字符 H 出现的次数
l.count('H')
```

```python
# 统计字符 l 出现的次数
l.count('l')
```

**Point:** **<span style="color:orange">字符串对象 vs 列表对象</span>**

字符串对象与列表对象均可支持各自的 count 方法。

## 3. in 函数

&emsp;&emsp;in 函数，用于判断某元素是否出现在列表中，并返回布尔类型对象结果。

```python
l = list(range(4))
print(l)
```

```python
# 判断 0 是否在 l 中
0 in l
```

```python
# 判断 4 是否在 l 中
4 in l
```

## 4. sort 方法

&emsp;&emsp;sort 方法主要用于列表的排序，尤其是由数值型对象组成的列表排序。

```python
l = [1, 0, 2, 3]
print(l)
```

```python
# sort 方法将在原对象的基础上进行修改
l.sort()
```

```python
# 查看修改后的结果
l
```

理解：在 REPL 环境中，每个可以执行的表达式都必将返回一个结果。返回结果要么赋值给别的变量，要么打印到屏幕上然后释放内存，若表达式中既没有赋值过程，也没有最终在屏幕中显示结果，则表明表达式的结果替换了原变量所指代的对象。

```python
l = [1, 0, 2, 3]
```

```python
# sort 方法有一个 reverse 参数，当参数取值 True 或 1 时，进行降序排列。
l.sort(reverse=True)
```

```python
# 查看最终降序排列结果
l
```

## 5. reverse 方法

&emsp;&emsp;reverse 方法用于对序列的元素排列进行翻转排序

```python
l = [1, 0, 2, 3]
```

```python
# reverse 方法，对序列进行翻转
l.reverse()
```

同样，reverse 方法直接在原对象上进行修改

```python
l
```

```python
l = list("Hello")
l.reverse()
print(l)
```

**Point:**

- reverse 方法只是将列表对象中的元素逆向排列。
- sort 方法不能对不同对象类型的元素进行排序，默认行为是升序排列，而使用 reverse=True 参数时，即为降序排列。
