&emsp;&emsp;布尔型（bool）对象也被称为逻辑型对象，布尔对象类型只包含两个值，分别是 TRUE 和 FALSE，主要用于进行逻辑运算或者表示某种逻辑运算结果。

## 1. 创建布尔对象

```python
# 首字母大写，Python 区分大小写。
True
```

```python
type(True)
```

## 2. 比较运算

比较常见的生成布尔型对象的表达式就是比较运算。常用比较运算的运算符如下所示：

| 运算符 | 描述 | 示例 |
| :------| :------ | :------ |
| == | 检查两个操作符的值是否相等，如果是，则返回结果为真 | 例如 a=3，b=3，则（a == b）为 True |
| != | 检查两个操作符的值是否相等，如果不相等，则返回结果为真 | 例如 a=1，b=3，则（a != b）为 True |
| > | 检查左操作数是否大于右操作数的值，如果是，则返回结果为真 | 例如 a=7，b=3，则（a > b）为 True |
| < | 检查左操作数是否小于右操作数的值，如果是，则返回结果为真 | 例如 a=7，b=3，则（a < b）为 False |
| >= | 检查左操作数是否大于或等于右操作数的值，如果是，则返回结果为真 | 例如 a=3，b=3，则（a >= b）为 True |
| <= | 检查左操作数是否小于或等于右操作数的值，如果是，则返回结果为真 | 例如 a=3，b=3，则（a <= b）为 True |

```python
# 恒等操作符
a = 3
b = 3
a == b
```

```python
a = 1
b = 3
a != b
```

```python
a = 7
b = 3
a > b
```

```python
a >= b
```

## 3. 逻辑运算

对于布尔型对象，还有一类独有的运算：逻辑运算。布尔型的逻辑运算与数学意义的逻辑运算一样，主要由最基础的与、或、非三种运算构成。当然，在这三个基础运算之上，还能够构建更加复杂的逻辑运算。

| 运算符 | 逻辑表达式 | 描述 |
| :------ | :------ | :------ |
| and | x and y | “与” 运算：如果 x 为 False，x and y 返回 False，否则它返回 y 的计算值。|
| or | x or y | “或” 运算：如果 x 是 True，它返回 True，否则它返回 y 的计算值。|
| not | not x | “非” 运算：如果 x 是 True，返回 False。如果 x 为 False，返回 True。|

理解：and、or 和 not 都是运算符，用于连接布尔型对象，组成逻辑运算表达式。

```python
True and True
```

```python
not False
```

```python
# 更加复杂的逻辑判别表达式
(1 < 2) or (1 > 2)
```

## 4. 布尔型对象的数值对象的对应关系

在 Python 和其他很多编程语言中，为了方便某特定场景的计算，数值对象 0、1 会分别和 False、True 分别对应。但这种对应是有限程度的对应，并不是完全一样。

对应关系主要体现在以下两个方面：

1. 对应关系体现在恒等时为 True

```python
1 == True
```

```python
0 == False
```

2. 对应关系还体现在可参与逻辑运算

```python
1 or False
```

```python
not 0
```

**对象类型的不同**：但本质上数值型对象和布尔型对象属于不同对象，虽然有对应关系，但本质上仍然不同，可以用 is 来判断对象类型是否一致。

```python
1 == 1.0
```

```python
import warnings, keyword

warnings.filterwarnings('ignore', category=SyntaxWarning)  # 使用 is 关键字判断不同对象类型时可能返回 SyntaxWarning，因此关闭警告。
keyword.kwlist  # 打印 Python 内置的关键字
```

```python
1 is 1.0
#print(type(1))
#print(type(1.0))
```

```python
1 == True
```

```python
1 is True
```

```python
1 is 1.0
```
