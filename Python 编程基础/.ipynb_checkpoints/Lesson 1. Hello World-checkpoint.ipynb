什么是对象？<br>

## 2. 函数对象 print 基本功能

```python
# 打印单个值
print(1)
```

```python
# 打印多个值
print(1, 2, 3)
```

理解：当 print 函数打印多个值时，实际相当于传入了多个参数依次打印。

```python
# 打印多个不同类型的值
print(1, "Hi")
```

```python
# 打印多个值，并且用逗号对所打印的内容进行分割。
print("I", "Robot")
print("I", "Robot", sep=',')
#print(sep=',', "I", "Robot")  # sep 参数只能置于最后，否则报错。
```

**Point: print 函数中的 sep 参数**

- sep 是 print 函数中的常用参数，并且是一个带有参数名的参数，在调用的过程中需要标注参数名；
- sep 参数的作用是，当 print 函数需要打印多个对象时候，输出结果将以 sep 输入参数对象进行分割。

```python
print("Hello", "World", sep=',')
```

## 3. Python 赋值

**Point:**

- Python 中常用等号进行赋值，赋值的本质是将对象赋给变量；
- 在等号赋值的过程中，等号左边是变量，右边是对象；
- Python 中对象赋值变量也可理解为给对象起名，然后方便对对象进行调取；
- 当然，既然是取名，也必然有自己的规范，这部分规范也就是变量命名规范。

```python
# 用等号进行赋值
a = "Hello"
```

理解：在上述赋值语句中，a 为变量，"Hello" 为字符串对象，通过等号将字符串对象 "Hello" 赋给 a 变量，后续调用 a 对象就等价于调用字符串 "Hello" 对象。

```python
a
```

```python
print(a)
```

```python
print("Hello")
```

```python
b = 1
```

```python
print(b)
```

```python
c = 2
```

```python
b + c
```

```python
# 函数也可被赋值
p = print
```

```python
p
```

```python
p(a)
```

```python
print("Hello")
```

理解：如上所述，Python 中一切皆对象，函数也是对象，也可以赋值给变量。

**删除对象赋值关系**

```python
# 新赋值代替旧赋值
p = 1
```

```python
p
```

```python
# 使用 del 删除赋值关系
del p
```

```python
p
```

## 4. 变量命名规则

"取名不规范，领导两行泪"<br>

**Point:**<br>

- 变量名的长度不受限制，但其中的字符必须是字母、数字、或者下划线，而不能使用空格、连字符、标点符号、引号或者其他字符；
- 变量名的第一个字符不能是数字，而必须是字母或者下划线；
- 变量名不区分大小写；
- 不能将 Python 关键字用作变量名。

```python
# 次单词首字母大写
kafkaTest = 1
```

```python
# 各单词用下划线分割
kafka_test = 2
```

```python
print(kafkaTest, kafka_test)
```

```python
# print 为内部保留关键字（内置函数名），用于充当变量会使原功能失效。
print = 1
```

```python
print
```

```python
print("Hello World!")
```

```python
# Alberthua 新增：删除赋值关系
del print
```

```python
import keyword

# 打印 Python 的关键字列表
print(keyword.kwlist)
```

```python
a = 1
b = 2
```

```python
a
```

**Point: 赋值不是相等**

- 变量只是指向了一个内存地址，内存地址中存储了对象；
- 变量可以重复赋值，后面的值会覆盖前边的值；
- 改变变量的值，其实就是改变了变量指向的内存地址。
