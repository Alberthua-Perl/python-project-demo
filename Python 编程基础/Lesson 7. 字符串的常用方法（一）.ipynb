## 1. 什么是方法？

在 Python 中，函数和方法是两个不同的概念。从使用者角度而言，假设 **Object** 代表某对象，**Function** 为函数，**Method** 为方法，则函数的使用方法为 **`function(object)`**，方法的使用方法为 **`object.method()`**。例如，继续上一节的问题，当需要替换 'Hallo' 中的 'a' 为 'e' 时，就可以使用字符串的一个常用方法：`replace`。

```python
# 创建字符串对象
a = "Hallo"
print(a)
```

```python
# 使用 replace 方法替换其中的 a 为 e
a.replace('a', 'e')
```

**Point:**

- object.method() 本身也是一个 **表达式**，且方法内部也可传入参数；
- **由上述结果也可知，该表达式并不会针对原对象进行修改，而是返回了一个新的修改后的对象；**
- replace 方法的参数解释，第一个参数为需要替换的对象，第二个参数为替换后的对象。

```python
a
```

```python
# 将修改完的对象赋值给b
b = a.replace('a', 'e')
print(b)
```

**Q&A:** 如何获取方法的帮助？以 replace 为例

```python
a.replace?
```

```python
# 查看方法的帮助，不能直接使用对象。
"Hi".replace?
```

```python
"Hi".replace
```

replace 方法的可选参数作用是，但出现多个可选的替换对象时，选择替换其中前几个。

```python
a
```

```python
# replace 默认全部替换
a.replace("l", "a")
```

```python
# a 中包含两个 l，设置参数，替换几个相同的字符。
a.replace("l", "a", 1)  # 替换一个相同的字符
```

```python
# a 中包含两个 l，此处替换前两个 l（和不输入参数时相同）
a.replace("l", "a", 2)
```

## 2. 字符串的常用方法

### 2.1 find 方法

&emsp;&emsp;用于在指定的区间内寻找确定字符串的（**起始位置**）索引值。

```python
# 在字符串 Hello 中寻找 H 字符的索引值
"Hello".find('H')
```

```python
# find 过程其实就是索引的逆过程
"Hello"[0]
```

```python
# 在字符串 Hello 中寻找 e 字符的索引值
"Hello".find('e')
```

```python
# 对于多次出现的字符，将返回该字符第一次出现的索引。
"Hello".find('l')
```

```python
# 查看 find 方法帮助文档
a.find?
```

**场景1：**

```python
# 字符串区间索引：返回整个字符串中在索引3与5之间该字符第一次出现的索引
"Hello".find('l', 3, 5)
```

```python
# find 方法可指定搜索范围，此处表示在 [0, 2) 索引区间内搜索 'l'，若无结果，则返回 -1。
"Hello".find('l', 0, 2)
```

**场景2：**

```python
# 字符串切片索引：先做切片，再从切片中做索引。
"Hello"[3: 5].find('l')
```

```python
# 先切片后索引与先索引后切片
"Hello"[0: 2].find('l')
```

**Point:** 以上两种场景存在区别，第一种从整个字符串中索引，第二种从切片中索引。

### 2.2 count 方法

&emsp;&emsp;用于计算指定索引区域内特定字符出现的次数。

```python
# 参数只给出字符时，会在字符串全区域进行计数。
"Hello".count('l')
```

```python
# 在索引区间 [2, 3) 中计算 l 出现的次数
"Hello".count('l', 2, 3)
```

### 2.3 split 方法

&emsp;&emsp;以指定字符切分字符串，切分过程为，以该字符为界，字符前组成一个字符串，字符后组成一个字符串，并删除切分字符。

```python
# 将 Hello 对象按 e 字符进行切分，e 之前、e 之后分别组成两个新的字符串，返回新字符串组成的列表。
"Hello".split('e')
```

提示，切分结果返回的是一个序列对象，序列对象将在下一节进行讲解。

```python
# 两个连续字符作为切分依据时，将出现空字符。
"Hello".split('l')
```

```python
# 首字母或尾字母字符作为切分依据时，将出现空字符。
"Hello".split('H')
```

```python
"Hello".split('o')
```

```python
# 示例：将切分的字符串重新拼接为字符串（Alberthua 新增）
src = "Happy learning aiops!"
output = ""

for item in src.split(' '):
    output = output + item + ' '

print("Source string is: ", output.strip())
```
