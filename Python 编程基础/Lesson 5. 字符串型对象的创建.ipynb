### 1.2 单引号和双引号

单引号和双引号都可用于创建字符串，但在具体的应用场景中，却各有用途。核心问题在于，若想创建一个内部带有引号的字符串，应该如何操作呢？

```python
# 若字符串内部包含双引号，则可以在外部使用单引号对字符串进行标注。
'"Hello World!", he said.'
```

```python
# 若字符串内部包含单引号，则可在外部使用双引号对字符串进行标注。
"What's up?"
```

但若要创建一个 "What's Up?",he said 字符串，应该如何操作？

```python
# 此时，考虑使用转义符，对内部不用标注字符串的引号进行转义。
'"What\'s Up?", he said'
```

```python
print('"What\'s Up?", he said')
```

### 1.3 字符串换行

很多时候，我们需要在一个字符串内进行换行操作，一般来说，字符串内的换行可以有两种方法：

- 其一，是使用三组引号（单引号或双引号均可）来标注带有换行的字符串：

```python
# 三组引号内的换行操作符将被识别
"""Hello!
Hi~"""
```

```python
'''Hello
Hi!~'''
```

```python
print("""Hello!
Hi~""")
```

```python
print('''Hello!
Hi!~''')
```

- 其二，则是使用 \n 作为换行符：

```python
"Hello!\nHi~"
```

```python
print("Hello!\nHi~")
```

**Point：理解转义符**

- Python 中转义符为 \，顾名思义，其核心用途就是转换对应符号原有的意义；
- \n 代表将原来字符的 n 的意义及逆行转化，转化后不再代表一个单独的字符，而代表换行；
- \' 代表将原来字符的 ' 的意义进行了转化，转化后不再代表用于标注字符串的引号，而仅仅代表字符串引号（仅表示字符字面义）。
