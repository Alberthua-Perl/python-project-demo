索引切片方面，列表和字符串完全一致。

```python
ls = list(range(6))
print(ls)
```

```python
# 按索引值进行索引
ls[1]
```

```python
# 按索引值进行切片
ls[1: 3]
```

```python
# 索引值的负数代表反向索引
ls[-1]
```

```python
# 仍然是包含三个索引参数
ls[1: 4: 2]
```

```python
# 反向索引奇数
ls[::-2]
```

和字符串不一样的地方在于，字符串不能通过索引对其内部部分内容进行修改，但列表可以。

```python
a = "Hello"
a[1] = 'a'
```

```python
l = list(range(3))
print(l)
```

```python
l[0]
```

```python
# 尝试对部分元素进行修改
l[0] = 1
print(l)
```

在 Python 中，这种内部不可修改的对象又被称为不可变对象，内部元素可修改的对象也被称为可变对象。字符串是不可变对象，而列表则是可变对象。关于可变或不可变，后续我们还将进行更深入的讨论。

和字符串类似，列表对象也有非常多的常用方法，用于对列表内部元素进行调整。

## 1. 列表运算符

&emsp;&emsp;列表也有运算符，使用运算符对列表进行修改，是最简单的列表操作方法。和数值型对象不同，列表对象只支持加法和乘法两种运算符。

- 加号运算符：简单的将各列表元素进行合并，组成新的列表。

```python
l1 = [1, 3, 5, 7, 2]
l2 = [2, 4, "Hi"]
```

```python
# 尝试进行列表 “相加”
l1 + l2
```

- 乘号运算符：列表元素合并后顺序不变，且重复元素会被保留。

```python
# 列表乘以某数
l2 * 2
```

列表相乘相当于先复制后拼接。

## 2. append 方法

&emsp;&emsp;append 方法主要用于对列表对象进行追加元素操作。

```python
l = list(range(3))
print(l)
```

```python
# 在列表的末尾，追加数值对象 3。
l.append(3)
```

```python
l
```

```python
# 可以追加任意类型的对象，比如追加列表对象。
l.append([3, 2, 1])
```

```python
l
```

```python
# 追加函数对象
l.append(print)
```

```python
print(l)
```

## 3. extend 方法

&emsp;&emsp;append 方法一次只能追加一个对象，若要一次追加多个对象，则需要使用 extend 方法。当使用 extend 方法时，其参数位是需要添加的多个对象元素共同组成的列表，即需要被追加的多个对象必须装在一个列表容器中，然后传入 extend 方法。

```python
l = list(range(4))
print(l)
```

```python
# 在 l 中追加两个数值型对象 1、2
l.extend([1, 2])
print(l)
```

```python
# 当然，extend 函数也可以只追加一个元素，但即便是一个元素，也应当将其置于一个列表容器中。
l.extend([1])
print(l)
```

```python
# 尝试追加一个列表元素，这时候需要使用列表的嵌套。
l.extend([[1, 2, 3]])
print(l)
```

```python
# 和 append 一样，extend 也可以追加不同类型对象
l.extend(["Hi", print])
print(l)
```

## 4. insert 方法

&emsp;&emsp;append 和 extend 只能在列表的末尾追加，若想要在任意位置添加元素，则需要使用 insert 函数。insert 函数和 append 函数一样，一次只能添加一个元素，并且在定位位置的时候，还是需要索引值的帮助。

```python
l = list(range(3))
print(l)
```

```python
# insert 方法第一个参数指代插入元素的索引值，第二个元素用于指代插入的元素对象。
l.insert(1, [2, 3])
print(l)
```

## 5. pop 方法：索引值删除

&emsp;&emsp;和上述添加元素的方法正好相反，pop 方法主要用于删除列表中的指定元素。

```python
l = list(range(5))
print(l)
```

```python
# 默认情况下（不输入任何参数的情况下），pop 会删除列表中最后一个元素，并返回该元素。
l.pop()
```

```python
# 删除后的结果
l
```

```python
# pop 方法也可以通过输入索引值来删除列表中的指定元素
l.pop(0)
```

```python
# 删除后的结果
l
```

## 6. remove 方法：列表值删除

&emsp;&emsp;pop 方法通过索引值定位需要删除的对象，也被称为 **索引值删除**。除此之外，实际应用当中，经常还需要依据具体的元素内容来进行删除，通过指定删除内容来执行删除，此时就需要用到 remove 方法。

```python
# 转换字符串对象为列表对象
l = list("Helalo")
print(l)
```

```python
# 在列表中删除 H 字符对象
l.remove('H')
```

```python
# 删除之后的结果
l
```

```python
# 删除字符 l，但注意，l 出现了两次。此时 remove 函数会删除第一次出现的 l。
l.remove('l')
```

```python
# 删除之后的结果
l
```

## 7. del 函数

&emsp;&emsp;del 函数是 Python 中最常用的删除函数，其本身不是列表的特有方法。使用 del 函数，可依照索引切片直接删除列表中的部分函数，也可删除整个对象。

```python
l = list(range(5))
print(l)
```

```python
# 删除 l 中索引值为 2 的元素，del 函数基本使用方法是 del 后接要删除的元素或对象。
del l[2]
```

```python
# 删除后的结果
l
```

```python
# 删除一个切片
del l[1:]
```

```python
l
```

```python
# 删除整个列表对象，也可以是其他对象。
del l
```

```python
l
```

## 8. clear 方法

&emsp;&emsp;若要一次性清空列表对象中的全部元素，可使用 clear 方法。

```python
l = list(range(5))
print(l)
```

```python
# 清空列表
l.clear()
```

```python
# 返回空列表
l
```
