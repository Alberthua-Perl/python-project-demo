索引是 Python 中非常重要的概念，借助索引，我们可以对字符串内部进行灵活修改。 

## 1. 字符串索引

Q：但方向某字符串对象 "Hallo" 语法错误，想将其修改为 "Hello"，该如何操作？

```python
a = "Hallo"
```

A：基本思路是，先定位到错误的字符 a，然后再进行修正

```python
# 利用索引，定位错误字符。索引的本质就是根据某种规则进行定位和提取。
a[1]
```

**Point: Python 索引基本规则**

- 对字符串进行索引时，基本格式为 **字符串[索引值]**；
- Python 的索引值是从 0 开始进行计算，也就是说，字符串中的第一个字符对应索引值为 0，第二个字符对应索引值为 1，以此类推。

```python
# 索要索引第三个字符
a[2]
```

```python
# 注意：字符串不可通过索引进行修改
a[1] = 'e'
```

因此，关于 Python 索引，我们大概了解其功能，就是利用索引，对原始对象进行有针对性的信息提取。并且字符串是支持索引的，但不能直接对索引出来的结果通过赋值的方式进行修改。此处至于如何修改，我们将进一步讨论。

## 2. 反向索引

除了上述索引方法外，对于字符串而言，从最后一个字符往前，索引值还可以分别取为 -1、-2、-3、... 以此类推。

```python
# 索引最后一个字符
a[-1]
```

```python
# 索引倒数第二个字符
a[-2]
```

## 3. 字符串切片

切片和索引是伴生的概念。所谓切片，指的是利用类似索引的方法，按照某种规则，一次性提取字符串的一个片段的字符。

```python
a = "零一二三四五六七八九十"
```

```python
# 索引，挑选其中某个字符。
a[0]
```

```python
# 切片，挑选其中某片段字符
a[0: 2]
```

```python
a[2:-2:3]
```

**Point: 切片规则**

- 但索引中括号内出现冒号时，则默认进行切片运算；
- 切片的基本规则是：**字符串[切片起始位置:切片结束位置:每隔多少取个数]**，即 **string[start:end:offset]**；

> 1. 偏移量（offset）可以是正数（从前往后索引），也可以是负数（从后往前索引）。<br>
> 2. **当 start > end 时，索引返回空，除非指定 offset。**

- 在默认情况下，切片起始位置为 0，结束位置值为字符串长度-1，第三个参数默认值为1（偏移量），即全部取到；
- Python 切片，左闭右开（左包含右不包含）。

```python
# 不出现冒号进行索引
a[1]
```

```python
# 出现冒号时进行切片
a[1: ]
```

```python
# 右边界取 10 时，相当于对前 10 个数进行索引，并不包含第11个数值
a[1: 10]
```

```python
# a[1: ] 等价于 a[1: 11]
a[1: 11]
```

```python
# 每隔两个数取一个 
a[1: 10: 2]
#a[1:10:2]
```

```python
# 取奇数位
a[: : 2]
```

和索引类似，切片时其实索引值和末尾索引值也可用负号标注。

```python
a
```

```python
# 第一个元素负数索引值为 -11，最后一个元素索引值为 -1，左闭右开。
a[-11: -1]
```

```python
# 若起始位置索引大于末尾索引值，则会返回空值。
a[-1: -11]
```

```python
# 切片第三位参数也可以取负数，表示从后往前索引。
# 具体取值为 -1 时候，表示从后向前依次索引。
a[: : -1]
```

```python
# 具体取值为 -2 时候，表示从后向前每隔一个元素取一个，依次索引。
a[: : -2]
```

```python
# 第三位取为负时，第一位索引值（start）必须大于第二位索引值（end）。
a[11: 1: -1]
```

```python
# 未指定偏移量，第一位索引值大于第二位索引值，返回为空。
a[11: 1]
```

```python
# 偏移量 -1 为从后往前索引，而第一个索引值大于第二个索引值为从前往后索引，因此返回为空。
a[1: 11: -1]
```
