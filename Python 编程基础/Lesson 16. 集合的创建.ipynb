&emsp;&emsp;集合是一类非常特殊的对象，其既有前面提到的列表、元组等容器的功能特点，是一种 **无序且可变类型容器**，但只用简单形象的容器角度理解集合又会有所欠缺。一个比较合适的理解集合的角度，是将其看作字典的一种特殊形式 —— **由字典的 keys 所组成的列表**。由此我们可知，集合内的元素和字典中的 keys 一样，其一是必须为不可变对象类型，其二是元素之间不能重复。

## 1. 基本方法

&emsp;&emsp;和其他容器创建过程一样，集合也有一个一般创建方法，使用 `{}` 进行标注。

```python
{1, 2, 3}
```

```python
# 无序性体现
{1, 2, 'Hi', (1, 2, 3)}
```

&emsp;&emsp;字典 key 值特性：

```python
# 集合中不能出现可变对象
{{}, 1}
```

```python
d = dict([('a', 5), ('b', 9)])
d
# 集合中包含字典对象变量而报错
{d, 'hello'}
```

```python
# 集合中不可出现重复对象
{1, 2, 3}
```

```python
type({1, 2})
```

**Points:** 集合中的每个元素都必须是不可变对象！

## 2. set 函数创建集合

&emsp;&emsp;当然，除了一般创建法，我们还可以用 set 函数来创建集合。

```python
# list 转 set
set([1, 2, 3])
```

```python
# set 转 list
list({1, 2, 3})
```

```python
# tuple 转 set
set((1, 2, 3))
```

```python
# set 转 tuple
tuple({1, 2, 3})
```

```python
# str 转 set（同时去重）
set('Hello')
```

```python
# range 转 set
set(range(5))
```

```python
d = dict(A = 1, B = 2, C = 3)
print(d)
```

```python
d.keys()
```

```python
set(d.keys())
```

这里需要注意的是，集合转化为列表或元组，本质上都是 **无序变量** 转化为 **有序变量**，此过程需要注意输入顺序和最终输出顺序的区别。

```python
# 无序变量生成过程会自动跳转顺序
{1, 3, 2}
```

```python
# 最终输出顺序实际上已经发生了改变
list(set([1, 3, 2]))
```

注意：由于 {} 表示空字典，因此若要创建空集合，则必须借助 set 函数。

```python
type({})
```

```python
s = set()
print(s)
```

```python
type(s)
```

## 3. 利用集合去重

&emsp;&emsp;集合的不可变包含重复值的特性，使得我们可以借助集合来进行去重操作。

```python
l1 = [1, 1, 2, 3]
```

```python
s1 = set(l1)
s1
```

```python
ls1 = list(s1)
ls1
```

```python
l2 = [2, 3, 1, 1]
```

```python
# 去重也需要注意顺序调整
s2 = set(l2)
s2
```

```python
ls2 = list(s2)
ls2
```
