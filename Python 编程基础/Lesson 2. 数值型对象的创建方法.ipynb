## 2. 创建浮点型对象

简单理解，Python 浮点型对象就是数学意义中带小数位数的对象。

```python
b = 3.14
```

```python
b
```

```python
type(b)
```

Python 中用 float 指代浮点型对象，我们也可以使用同名函数创建浮点型对象。

```python
float(1)
```

```python
# 使用科学计数法创建浮点型对象
c = 1.23e2
```

```python
c
```

```python
type(c)
```

在科学计数法中，e 代表以 10 为底，e 后面数字代表 10 的幂。  
例如：1.23e2 代表 1.23 * 100 = 123

**Point:**

- 通过科学计数法创建的数，即使是整数，也会用浮点型来表示；
- 浮点型也可用于存储整数数值。

```python
f1 = 1
```

```python
f2 = 1.0
```
