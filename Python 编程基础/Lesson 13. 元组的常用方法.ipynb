&emsp;&emsp;由于元组和列表的高度类似，绝大多数列表的方法都可作用于元组。当然，由于元组的不可变性，用于修改容器内元素的一些增删改查的方法除外。

## 1. 元组的运算符

&emsp;&emsp;和列表一样，元组也有加号和乘号两种运算符。

```python
t1 = (1, 3, 4)
t2 = (1, 'Hi')
```

```python
t1 + t2
```

t2 * 2

## 2. index 方法

&emsp;&emsp;根据内容，查找索引。

```python
t = tuple(range(4))
print(t)
```

```python
t.index(3)
```

## 3. count 方法

&emsp;&emsp;用于计算元组内某元素出现的次数。

```python
t = tuple(range(4)) + (1, 3)
```

```python
print(t)
```

```python
t.count(3)
```
