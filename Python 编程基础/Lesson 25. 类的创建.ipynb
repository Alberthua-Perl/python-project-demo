## 3. 类的创建

&emsp;&emsp;接下来，我们就开始尝试着进行类的创建。前面说了，类的创建就是在 Python 的世界中创造一类新的元素，此处我们以 “人”为例，尝试如何在 Python 中创建“人类”。

### 3.1 类的定义与实例化方法

&emsp;&emsp;在 Python 中，我们使用 **class 关键字** 来定义类，基本格式和定义函数类似。

```python
class Human:
    '''人类'''
    pass
```

&emsp;&emsp;此处我们就完成了一个名为 Human 的类的创建，需要注意的是：

- 在类中，我们使用三引号来标注帮助信息；
- 如果定义的类没有任何内容，可以使用 pass 来完成类的创建；
- 类的名称要求第一个单词开始，每个单词首字母大写，也就是所谓的驼峰式命名。

```python
# 查看 Human 类的说明
help(Human)
```

&emsp;&emsp;然后再来看类的实例化过程：

```python
# Human 实例化：张三
zs = Human()
```

```python
# 能够看出，zs 是 Human 的实例化对象。
zs
```

&emsp;&emsp;但由于我们没有对 Human 类有任何说明，因此暂时无法对 zs 对象进行任何操作。

### 3.2 类的定义过程

&emsp;&emsp;我们之前说，类是某一些对象的共性的总结。简单来说，在 Python 定义类的过程中，总共可分为两类内容，以 Human 为例：

- 其一，是需要给出这类对象共同的 **静态类型属性**，例如，都是哺乳类动物；
- 其二，则是需要给这类对象保留一些实例化过程中，能够表现特性的 **属性或方法的接口**，例如，每个人的年龄、身高、体重可能各有不同，你可以规定 Human 类有共同的哺乳动物这种属性，但你无法规定所有 Human 的年龄、身高、性别。

#### 3.2.1 定义静态属性

```python
class Human:
    '''人类'''
    mam = True    # 是哺乳类动物
```

```python
zs = Human()
```

```python
zs.mam            # 查看 zs 是否是哺乳类动物
```

```python
ls = Human()      # 查看 ls 是否是哺乳类动物
ls.mam
```

**Points：**

- 此处 mam 就是所有 Human 类的属性，每个实例化的类都拥有该属性，并可以通过 `.属性` 的方法进行查看；
- mam 不是单独存在的全局变量，虽然在类的外部可以调用，但只能通过具体对象属性的方法来调用。

```python
mam
```

#### 3.2.2 定义动态属性

&emsp;&emsp;形象解释，例如对于某些情况下，我们需要在实例的过程中说明对象的某些属性，这些属性往往是因对象不同而改变的属性，例如年龄、性别等，这些属性 Human 类都有，但具体取值又有所不同，类似于此前说的，对象的值的部分。

```python
class Human:
    '''人类'''
    mam = True
    def __init__(self, age = 0, gender = 'Unknow'):    # 设置动态属性默认值
        self.age = age
        self.gender = gender
```

&emsp;&emsp;**`__init__`** 函数解释：

- 是一种特殊函数，也称为是 **构造函数**，多用于带参数属性的类的创建过程；
- self 是一个特殊的参数，相当于类本身的一种表示，是 `__init__` 函数必须设置的第一个参数。

&emsp;&emsp;类的改写完成后，我们就能尝试实例一个 Human：

```python
zs = Human(10, 'M')    # 重新设置属性
```

```python
zs.age                 # 对比 __init__ 函数中 self.age
```

```python
zs.gender              # 对比 __init__ 函数中 self.gender
```

```python
ls = Human()
```

```python
ls.age                 # 相当于 __init__ 函数默认取值的返回结果
```

```python
ls.age = 10
```

```python
# 属性的修改
ls.age
```

&emsp;&emsp;**`__init__`** 函数的其他说明：

- 它是特殊函数，名称不可修改；
- 两个下划线表示该函数是 **私有函数**，只允许类的内部访问，无法通过外部访问。

```python
class Human1:
    '''人类'''
    mam = True 
    def __began__(self, age = 0, gender = 'Unknow'):
        self.age = age
        self.gender = gender
```

```python
zs1 = Human1(10, 'M')
```

```python
zs1 = Human1()
```

```python
zs1.mam
```

```python
zs1.age
```

&emsp;&emsp;修改完名称的 began 函数是什么函数？我们将在下一节进行讨论。

```python
class Human:
    '''人类'''
    mam = True 
    def __init__(self, age = 0, gender = 'Unknow'):
        self.age = age
        self.gender = gender
```

```python
# 只能在类的内部调用
__init__()
```

&emsp;&emsp;更进一步来说，首尾双下划线的类是表示定义的特殊方法，一般是系统定义的名字，无法外部访问，也不允许修改函数名称。

### 3.3 类的属性私有化

&emsp;&emsp;在此前的定义中，Human 的 mam 属性是可以通过外部进行修改的。

```python
zs.mam
```

```python
zs.mam = False
```

```python
zs.mam
```

&emsp;&emsp;在某些情况下，这么做是不合适的。因此，我们需要进一步掌握属性私有化的方法，或者说 **让类的属性无法在外部被修改** 的一般方法。

```python
# 利用 Python 装饰器
class Human1:
    '''人类'''
    mam = True          # 类的静态属性    
    @property           # Python 内置装饰器将实例方法转换为属性
    def mam(self):
        return False    # mam 属性值覆盖上方 mam 属性值
```

```python
zs = Human1()
```

```python
zs.mam
```

```python
zs.mam = True
```

&emsp;&emsp;**类的方法**：在定义类的过程中，如果内部出现定义的其他函数，则该函数就属于 **类对象的方法（实例方法）**。

```python
# 实例方法
class Human1:
    '''人类'''
    mam = True
    def AntiMam(self):
        return not self.mam
```

```python
zs = Human1()
zs.mam
```

```python
zs.AntiMam()    # 类实例化后对象的方法，返回 mam 逻辑运算相反的结果。
```

**Points:**

- 类中定义的函数也需要传入 self 参数；
- 这些函数可以通过外部调用方法的方式进行调用。

&emsp;&emsp;因此，我们可以理解，方法其实就是绑定到对象的函数，在定义的过程中，方法是在定义类的时候同步定义的，而函数则可以随时随地单独定义，同时我们也可以理解在定义类的过程中，__init__函数和一般函数区别：__init__函数定义的参数需要在类的实例化过程中给出，而一般函数则属于类型对象实例化之后的方法。

```python
# 如果属性和方法同名
class Human1:
    '''人类'''
    mam = True
    def mam(self):
        return True
```

```python
zs = Human1()
```

```python
zs.mam
```

```python
zs.mam()
```

如果属性和方法同名，将优先优先调用方法。而property，则可将方法转化为属性。因此，上述类的创建，相当于创建了一个属性mam和一个同名方法mam，并且将同名方法转化为了属性，而函数本身返回结果恒为True，因此mam属性为True就具备了不可修改性。

```python
class Human1:
    '''人类'''
    mam = True
    @property
    def mam(self):
        return True
```

```python
zs = Human1()
```

```python
zs.mam
```
