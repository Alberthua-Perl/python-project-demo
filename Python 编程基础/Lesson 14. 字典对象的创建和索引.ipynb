&emsp;&emsp;列表是有序的可变容器，元组是有序的不可变容器，而字典则是无序的可变容器。字典类型对象的元素组织和查找的方法都和现实中的字典完全一致，这也是该类型对象名称的由来。字典类型对象是 Python 开发中非常常用的对象类型。

## 1. 字典的创建   

&emsp;&emsp;Python 中的字典和真实环境中的字典类似，字典是由一个个键值对（Key-value）构成，键就相当于字典的目录，在键中存储有一个内存地址（相当于页码），然后我们在内存地址中查找到这个键对应的值（也就相当于这一个页面中存储的单词的详细讲解），字典也被称作 **关联数组** 或者 **哈希表**。

### 1.1 一般创建方法

&emsp;&emsp;字典由键（key）和对应值（value）成对组成的特殊元素构成，外侧则使用大括号圈定。

```python
d = {'A': 1, 'B': 2, 'C': 3}
```

```python
type(d)
```

**Point:**

- dict 即表示字典，其中 'A'、'B'、'C' 为字典中的键（key），并且键值一一对应，用冒号表示对应关系；      
- 一个字典可以包括多个键值对，每个键值对可看成是字典中的一个元素，每隔元素彼此之间用都好进行分隔；
- 字典中的 key 只能使用不可变类型对象充当，且一个字典中的 key 必须是独一无二的，而 value 可用任意类型对象；
- 🎯 不可变类型对象包括数字（整型数与浮点数）、字符串和元祖；
- 🎯 可变类型对象包括列表、字典和集合。

```python
# 尝试创建不同 key 类型的字典
d1 = {1: 1, 'A': 2, (1,): 1}
```

```python
print(d1)
```

当然，若把字典看成一种特殊的容器，字典理所当然可用创建空字典。

```python
dn = {}
```

```python
print(dn)
```

### 1.2 使用 dict 函数创建字典

&emsp;&emsp;和列表、元组类似，字典也可以用类型同名函数 dict 来进行创建。由于字典数据结构比较特殊，在使用 dict 函数创建的过程也较为特殊，dict 函数必须输入若干长度为 2 的元组组成的列表，才能顺利的转化为字典。

```python
dict([(1, 4), (2, 3)])
```

**Point:**

- 必须输入长度为二的元组组成的列表，在转化的过程中长度为 2 的元组中的第一个元素将变为 key，第二个元素将变为 value；
- 正是因为二元元组中第一个元素会对应转化为字典的 key，因此该元素对象类型必须是不可变类型对象，否则会报错。

```python
dict([([1, 2], 4), (2, 3)])
```

另外，dict 还有特殊用法，能将赋值过程转化为字典的对应关系。

```python
dict(a = 1, b = 2)
```

**Point：**

- 赋值过程转化为字典时，变量名称将会变成字典的 key，并且会转化为字符串；
- 利用该方法创建字典时，只能创建 key 为字符串型的字典。

## 2. 字典的索引

&emsp;&emsp;字典的无序特性导致字典的索引和字符串、列表以及元组都不一样。在实际使用中，我们是根据字典元素的二元性（key-value）进行索引，即通过输入 key 来查找对应的 value，而非利用索引值进行索引。

### 2.1 一般索引方法

&emsp;&emsp;基本索引格式为 dict\[key]，仍然是利用中括号进行索引，只是括号内不再是索引值而是要查找的 key 的名称。同时，索引返回结果也不是整个 key-value 元素，而是 key 对应的 value值。

```python
print(d)
```

```python
# 查找 key 为 A 的元素中的 value
d['A']
```

同时，由于字典是可变类型对象，因此索引出来的结果既可修改。

```python
d['A'] = 2
```

```python
print(d)
```

```python
d1
```

```python
d1[(1, )]
```

```python
d1[3]
```

### 2.2 get 方法

&emsp;&emsp;除了一般索引方法外，对于字典这类比较特殊的对象，通常还会考虑使用 get 方法来进行索引。和一般索引方法不同，get 方法可指定当对应的 key 不存在时返回的结果。

```python
d1
```

```python
# get 也是一种方法
d1.get((1, ))
```

```python
# 当对应 key 不存在时，不会返回任何结果。
d1.get(3)
```

```python
# 通过对第二个参数的值的输入，可指定输出当 key 不存在时返回的结果。
d1.get(3, 'Error')
```

```python
# 示例：一个简单的数据库（Alberthua 新增）
#
# 一个将人名用作键的字典
# 每个人都用一个字典表示，字典包含 'phone' 和 'addr'，它们分别与电话号码和地址相关联。
people = {
    'Alice': {
        'phone': '2341',
        'addr': 'Foo Drive 23'
    },
    'Beth': {
        'phone': '9102',
        'addr': 'Bar Street 42'
    },
    'Cecil': {
        'phone': '3158',
        'addr': 'Baz Avenue 90'
    }
}

# 电话号码和地址的描述性标签，供打印输出时使用。
labels = {
    'phone': 'phone bumber',
    'addr': 'address'
}

name = input("Name: ")
request = input("phone number (p) or address (a)? ")

key = request
if request == 'p':
    key = 'phone'
elif request == 'phone number':
    key = 'phone'
elif request == 'a':
    key = 'addr'
elif request == 'address':
    key = 'addr'

person = people.get(name, {})
#person = people[name]  #方法2
label = labels.get(key, key)
#label = labels[key]    #方法2
result = person.get(key, 'not available')
#result = person[key]   #方法2

print("{}'s {} is {}.".format(name, label, result))
```
