## 1. 变量作用域

&emsp;&emsp;还是从之前的一个例子讲起，假设一个函数的计算逻辑是让两个输入对象利用加号进行计算，则定义函数：

````python
def sum_1(a, b):
    a = a + b
    return a
````

````python
a = 1; b = 2
````

````python
print(a, b)
````

````python
sum_1(a, b)
````

````python
a
````

**理解：**

- 由此可知，此 a 并非函数中的 a。之前我们理解函数中的形参 a、b 为占位符，其实函数中的 a、b 是函数体内的 **局部变量**，其只能在函数中起作用，而我们另外单独定义的 a、b 则是 **全局变量**，可以在全域、跨不同函数中起作用，此二者虽然看起来名字相同，但其实是不同的对象，本质是因为 **作用域不同**。    

**Points:**

- 上述函数调用过程：本质上是将全局变量 a 指向的 1 赋给了函数体内的局部变量 a、全局变量 b 指向的 2 赋给了局部变量 b，局部变量 a、b 在执行函数计算过程中把各自指向的 1、2 相加后得出 3，再把 3 赋给局部变量 a，最终返回局部变量 a 所指向的 3，整个过程并不会修改全局变量的指向；
- 函数体内最终返回（return）的不是局部变量，而是局部变量指代的对象；
- 局部变量可以和全局变量重名，但在函数体内进行计算的过程中，优先会使用局部变量：

````python
sum_1(a = 2, b = 3)
````

````python
a
````

- 形参取值的过程，本质上就是给局部变量赋值的过程；
- **局部变量无法在除了函数内的其他地方使用，而全局变量可以在函数体内使用**：

````python
def sum_2(a, b):
    return a+b+c    # 变量 c 虽然在函数体内未定义，但作为全局变量可对其赋值。
````

````python
c = 1
sum_2(1, 2)
````

## 2. 全局变量的修改

&emsp;&emsp;那么接下来的问题就是，函数体内能否修改全局变量呢？对于可变类型全局变量，可以在函数体内对其进行修改，而不可变类型对象则不能直接修改。

- 可变类型全局变量在函数内修改，例如：对一个列表的索引值为奇数的值+1，并且将计算结果保存在另一个列表里面：

````python
l = [1, 2, 3, 4]
````

````python
lr = []

def add_1(l):
    for i in range(len(l)):
        if i % 2 != 0:
            lr.append(l[i]+1)
        else:
            lr.append(l[i])
````

````python
add_1(l)
````

````python
lr
````

````python
l
````

- 不可变类型全局变量不能直接修改，若一定要在函数内申明需要优先全局变量或对其进行修改，则需要使用 **global** 关键字：

````python
a = 1
````

````python
def sum_2(b):
    global a    # 若不使用 global 关键字声明，运行报错。
    a = a + b
    return a
````

````python
sum_2(2)
````

````python
a
````

值得注意的是，此时 a 就不能再作为局部变量名了。

## 3. lambda 函数

&emsp;&emsp;基本理解：lambda 函数又被称为 **匿名函数**，支持使用更加轻巧的方式定义轻便函数，例如：

````python
f1 = lambda x, y: x + y
````

````python
f1(1, 2)
````

本质上等价与：

````python
def f2(x, y):
    return x + y
````

````python
f2(1, 2)
````

其基本结构为：

```python
lambda [arg1 [, arg2, ... ]]:expression
```

当然，轻便的代价就是有诸多限制，例如，冒号后的表达式中不能出现 while 循环、表达式中如果要用 if 条件判别就一定要有 else 等；

在 lambda 函数中使用 if-else 判别：

````python
d = lambda x: x+1 if x > 10 else x - 1
````

````python
d(2)
````

````python
d(11)
````

等价于：

````python
def d1(x):
    if x > 10:
        return x+1
    else:
        return x-1
````

````python
d1(2)
````

````python
d1(11)
````
