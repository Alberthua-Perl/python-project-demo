## 2. 科学计算常用函数

除了基本运算符外，Python 还有更加复杂的科学计算函数，以下是 Python 自带的科学计算函数。

| 函数 | 描述 | 
| :------: | :------: | 
| abs(x) | 返回 x 的绝对值 | 
| max(n1, n2, ...) | 返回 n1、n2 最大值 | 
| min(n1, n2, ...) | 返回 n1、n2 最小值 |
| round(x[,n]) | 返回 x 的小数点后 n 位 |

```python
# 绝对值函数
abs(-1)
```

```python
abs(-9.57)
```

```python
# 最大值函数
max(2, 4, 1)
```

```python
min(2, 4, 1)
```

```python
round(3.14, 1)
```

```python
round(3.345, 0)
```

```python
round(3.345, 1)
```

```python
round(3.345, 2)
```

## 3. math 包中的其他科学计算函数

在 Python 的 math 包中，还有更多的科学计算函数。

```python
import math
```

math 包的常用计算函数如下：

| 函数 | 描述 |

| log10(x) | 返回以 10 为基数的 x 的对数 |
| log(x, y) | 返回以 y 为基数的 x 的对数 |
| pow(x, y) | 返回 x 的 y 次幂 |

```python
math.fabs(-1)
```

```python
math.fabs(-7.293)
```

```python
abs(-1)
```

```python
math.exp(2)
```

```python
# 求 10^x=100 的指数
math.log(100, 10)
```

## 4. 复合运算符

在实际数值运算过程中，我们常常需要将计算结果替换赋值给之前的变量，这个时候就需要使用复合运算符，相关运算规则如下：

| 函数 | 描述 | 实例 | 
| :------:| :------: | :------: | 
| += | 加法赋值运算符 | c += a 等价于 c = c + a | 
| -= | 减法赋值运算符 | c -= a 等价于 c = c - a | 
| \*= | 乘法赋值运算符 | c \*= a 等价于 c = c \* a |
| /= | 除法赋值运算符 | c /= a 等价于 c = c / a |
| %= | 取余赋值运算符 | c %= a 等价于 c = c % a |
| \*\*= | 幂赋值运算符 | c **= a 等价于 c = c ** a |
| //= | 加法赋值运算符 | c //= a 等价于 c = c // a |

```python
a = 6
```

```python
a += 3
a
```

```python
a -= 3
a
```

```python
a *= 3
a
```

```python
a /= 3
a
```

```python
a //= 3
a
```

```python
a %= 3
a
```

```python
a **= 3
a
```
