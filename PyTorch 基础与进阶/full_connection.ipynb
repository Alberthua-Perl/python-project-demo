```python
import torch
import torch.nn as nn # 用于定义神经网络
import torch.optim as optim # 用于定义优化器
from torchvision import datasets, transforms # 用于加载手写数字数据集
import matplotlib.pyplot as plt # 用于可视化
```

```python
# 数据预处理：将图片转换为 Tensor，并进行归一化（将像素值从 0-255 转换为 0-1）
transform = transforms.Compose([
    transforms.ToTensor(), # 转换为 Tensor，形状为[1, 28, 28]
    transforms.Normalize((0.1307,), (0.3081,)) # 归一化，均值和标准差是 MNIST 数据集的常用值
])

# 加载训练集和测试集
train_dataset = datasets.MNIST(
    root='./data', # 数据集保存路径
    train=True, # 表示加载训练集
    download=True, # 如果本地没有数据集，自动下载
    transform=transform
)
print("train_dataset 类型：", type(train_dataset))

test_dataset = datasets.MNIST(
    root='./data',
    train=False, # 表示加载测试集
    download=True,
    transform=transform
)
print("test_dataset 类型：", type(test_dataset))

# 数据加载器：将数据集分成批次（batch），方便模型训练
train_loader = torch.utils.data.DataLoader(
    train_dataset,
    batch_size=64, # 每批次加载 64 张图片
    shuffle=True # 训练时打乱数据，提高模型泛化能力
)
print("train_loader 类型：", type(train_loader))

test_loader = torch.utils.data.DataLoader(
    test_dataset,
    batch_size=64,
    shuffle=False # 测试时不打乱数据
)
print("test_loader 类型：", type(test_loader))
```

```python
class FullyConnectedModel(nn.Module):
    def __init__(self):
        super(FullyConnectedModel, self).__init__()
        # 定义第一层全连接（输入层→隐藏层）：输入 784 个特征，输出 128 个特征
        self.fc1 = nn.Linear(784, 128)
        # 定义 ReLU 激活函数
        self.relu = nn.ReLU()
        # 定义第二层全连接（隐藏层→输出层）：输入 128 个特征，输出 10 个特征（对应 10 个数字）
        self.fc2 = nn.Linear(128, 10)
        # 定义前向传播过程（模型的核心，数据从输入到输出的流动过程）
        
    def forward(self, x):
        # 将输入的图片 Tensor（形状为[batch_size, 1, 28, 28]）扁平化为 [batch_size, 784]
        x = x.view(-1, 784) # -1 表示自动计算 batch_size，784=28×28
        print(x.shape)
        # 输入层→隐藏层：线性变换 + ReLU 激活
        x = self.relu(self.fc1(x))
        # 隐藏层→输出层：线性变换（Softmax 激活在后续计算损失时统一处理）
        x = self.fc2(x)
        return x
        # 实例化模型

model = FullyConnectedModel()
# 查看模型结构
print(model)
```

```python
# 损失函数：分类任务使用交叉熵损失（CrossEntropyLoss），自动包含 Softmax 激活
criterion = nn.CrossEntropyLoss()
# 优化器：使用随机梯度下降（SGD），学习率为 0.01
optimizer = optim.SGD(model.parameters(), lr=0.01, momentum=0.9)
```

```python
# 定义训练轮次（epoch），这里设置为 5 轮
epochs = 5
# 训练过程
for epoch in range(epochs):
# 切换模型为训练模式（启用 Dropout、BatchNorm 等，本章暂未使用，养成习惯）
    model.train()
    # 初始化每轮的总损失
    total_loss = 0.0
    # 遍历训练集的每一个批次
    for batch_idx, (data, target) in enumerate(train_loader):
        # 1. 清零梯度（避免梯度累积）
        optimizer.zero_grad()
        # 2. 前向传播：输入数据，得到模型输出
        output = model(data)
        # 3. 计算损失
        loss = criterion(output, target)
        # 4. 反向传播：计算梯度
        loss.backward()
        # 5. 优化器更新参数
        optimizer.step()
        # 累加损失
        total_loss += loss.item()
        # 计算每轮的平均损失
        avg_loss = total_loss / len(train_loader)
        # 打印每轮的训练信息
        print(f'Epoch [{epoch+1}/{epochs}], Average Loss: {avg_loss:.4f}')

print("训练完成！")
```

```python
# 切换模型为评估模式（禁用 Dropout、BatchNorm 等）
model.eval()
# 初始化正确预测的数量和总样本数量
correct = 0
total = 0
# 测试过程中，不计算梯度（节省内存，加快速度）

with torch.no_grad():
    for data, target in test_loader:
        # 前向传播，得到模型输出
        output = model(data)
        # 得到预测结果（取输出中概率最大的索引，即预测的数字）
        _, predicted = torch.max(output.data, 1)
        # 累加总样本数和正确预测数
        total += target.size(0)
        correct += (predicted == target).sum().item()
        # 计算测试准确率
    
accuracy = 100 * correct / total
print(f'测试集准确率: {accuracy:.2f}%')
```
